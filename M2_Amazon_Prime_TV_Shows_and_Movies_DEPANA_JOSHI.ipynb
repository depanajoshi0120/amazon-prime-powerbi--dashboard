## 3. ***Data Wrangling***

### Data Wrangling Code

````python
import ast # parses source code into a hierarchical tree structure of objects
import pandas as pd

try:
    # Create copies for processing
    df_titles = titles.copy()
    df_credits = credits.copy()

    # 1. Handle Duplicates
    df_titles.drop_duplicates(inplace=True) #inplace=True (to modify the original DataFrame directly instead of creating a new copy)
    df_credits.drop_duplicates(inplace=True)

    # 2. Data Type Conversion
    def convert_to_list(col_val):
        if isinstance(col_val, str): #isinstance:if an object belongs to a specified class or data type # Checks if the incoming cell value is a piece of text/string
            try:
                return ast.literal_eval(col_val) # Safely converts the text into a real Python data structure
            except (ValueError, SyntaxError):
                return []
        return col_val

    df_credits['genres'] = df_credits['genres'].apply(convert_to_list)
    df_credits['production_countries'] = df_credits['production_countries'].apply(convert_to_list)

    # 3. Filling Missing Values
    # Movies dont have seasons so filling it with NaN
    df_credits.loc[df_credits['type'] == 'MOVIE', 'seasons'] = df_credits.loc[df_credits['type'] == 'MOVIE', 'seasons'].fillna(0) # Fills missing season counts with 0, but only for rows where the video type is a 'MOVIE'
    df_credits['imdb_score'] = df_credits['imdb_score'].fillna(df_credits['imdb_score'].median())
    df_credits['tmdb_score'] = df_credits['tmdb_score'].fillna(df_credits['tmdb_score'].median())
    df_credits['age_certification'] = df_credits['age_certification'].fillna('Unknown')

    # 4. Feature Extraction
    df_genres = df_credits.explode('genres') # Unpacks rows with lists in the 'genres' column into multiple rows, giving each genre its own separate line

    # 5. Filter Example
    df_top_rated = df_credits[df_credits['imdb_score'] >= 8.0]

    print(f"Cleaned Credits Shape: {df_credits.shape}")
    print(f"Missing Values in Score: {df_credits['imdb_score'].isnull().sum()}")

except NameError:
    print("Error: DataFrames 'titles' or 'credits' not found. Please ensure the loading cell (_uP18G2kyoHa) has been executed.")
````

````python
# Function to get mean & median for Amazon Prime content based on type (MOVIE/SHOW)
def get_mean_median(df, content_type): #This function returns the mean and median of numeric features for a specific content type (MOVIE or SHOW).


  try:
    # Filtering for the specific type and generating descriptive stats
    stats = df[df['type'] == content_type].describe()

    # Concatenating Mean (row 1) and Median (50%, row 5)
    return pd.concat([stats.iloc[1], stats.iloc[5]], axis=1).rename(columns={"50%": "median"}).fillna("-") #iloc selects rows by their numerical position index (starting at 0).
  except Exception as e:
    print(f"Invalid Type: {e}")

# Example usage:
# get_mean_median(df_titles, 'MOVIE')
# compares mean and median to see if the data is skewed
````

````python
# Creating impact metrics for high-rated titles with certification
# First, ensure the subset is defined from the cleaned credits data
df_high_rated_with_cert = df_credits[(df_credits['imdb_score'] >= 8.0) & (df_credits['age_certification'] != 'Unknown')].copy()

# Calculate performance metrics
df_high_rated_with_cert['popularity_per_min'] = df_high_rated_with_cert['tmdb_popularity'] / df_high_rated_with_cert['runtime']
df_high_rated_with_cert['votes_per_min'] = df_high_rated_with_cert['imdb_votes'] / df_high_rated_with_cert['runtime']
df_high_rated_with_cert['popularity_per_imdb_score'] = df_high_rated_with_cert['tmdb_popularity'] / df_high_rated_with_cert['imdb_score']
df_high_rated_with_cert['votes_per_tmdb_score'] = df_high_rated_with_cert['imdb_votes'] / df_high_rated_with_cert['tmdb_score']

print(f"Enhanced metrics calculated for {len(df_high_rated_with_cert)} certified high-rated titles.")
````
